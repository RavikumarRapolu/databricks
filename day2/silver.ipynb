```sql
%python
from pyspark.sql.functions import col
df= spark.table("dev.ravi_bronze.customer");
df2= spark.table("dev.ravi_bronze.transactions");
df = df.dropDuplicates().dropna().drop("ETL_TS");
df2 = df2.dropDuplicates().dropna().drop("ETL_TS").withColumnRenamed("customer_id","customer_id2");
df3 = df.join(df2, df.customer_id == df2.customer_id2,'inner').drop("customer_id2")

df3.write.mode("overwrite").saveAsTable("dev.ravi_silver.cust_trans");
```
