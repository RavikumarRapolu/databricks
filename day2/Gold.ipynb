```sql
%python

df= spark.table("dev.ravi_silver.cust_trans");

cust_spent = df.groupBy("customer_id").sum("amount").withColumnRenamed("sum(amount)","amount").orderBy("customer_id").withColumn("amount",round(col("amount")));
month_cat_spent = df.withColumn("month", date_format("transaction_date", "MMM")).groupBy("category","month").sum("amount").withColumnRenamed("sum(amount)","amount").orderBy("category","month").withColumn("amount",round(col("amount")));


cust_spent.createOrReplaceGlobalTempView("dev.ravi_gold.cust_spent")
month_cat_spent.createOrReplaceGlobalTempView("dev.ravi_gold.month_category_spent")

#cust_spent.write.mode("overwrite").saveAsTable("dev.ravi_gold.cust_spent")
#month_cat_spent.write.mode("overwrite").saveAsTable("dev.ravi_gold.month_category_spent")


```
